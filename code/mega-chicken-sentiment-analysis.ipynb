```python
"""
MEGA CHICKEN RESTAURANT - CORPORATE BRAND INTELLIGENCE DASHBOARD
Complete Single-Cell Script for Anaconda Jupyter Notebook
Run this entire block in ONE cell
"""

# ============================================================
# STEP 1: IMPORTS
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from collections import Counter
import string
import warnings
warnings.filterwarnings("ignore")

# ============================================================
# STEP 2: STYLE SETUP
# ============================================================
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['axes.edgecolor'] = '#333333'
plt.rcParams['axes.labelcolor'] = '#333333'
plt.rcParams['xtick.color'] = '#333333'
plt.rcParams['ytick.color'] = '#333333'
plt.rcParams['text.color'] = '#333333'
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Helvetica']
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 11
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10

COLORS = {
    'positive': '#2E7D32', 'negative': '#C62828', 'neutral': '#1565C0',
    'joy': '#FFD700', 'anger': '#D32F2F', 'trust': '#7B1FA2',
    'anticipation': '#FF8F00', 'sadness': '#455A64', 'fear': '#37474F',
    'surprise': '#00897B', 'warning': '#F57C00', 'info': '#1976D2',
    'dark': '#263238', 'light_grey': '#ECEFF1', 'cream': '#FFF8E1', 'navy': '#1A237E',
}

print("Step 1/5: Imports & Setup complete")
# ============================================================
# STEP 3: DATA LOADING
# ============================================================
raw_comments = [
    "My people", "Food for the Royal's.",
    "I've been following your restaurant for a while now and I have to say the food is always fresh and tasty always, keep up the good work",
    "See my people", "Mega chicken to the world", "My people", "Mega chicken to the world.",
    "Confirm I love this Caption In The Picture", "Lovely", "Mega chicken you people too much",
    "Superb,mega no dey carry last.", "Nice one.", "Very nice!", "Very interesting",
    "Yes we agreed with you but this remain for Lagos people",
    "I am looking for work in Mega chicken",
    "I love it delicious meal and juice.. Making me feel like I want more!!!!",
    "I like to have a taste of this delicious meal", "Happy Good Friday", "Exactly",
    "Happy New month", "My father in law 's restrant", "Mega yam", "Hmmmmm", "Yummy", "Nice", "Yummy",
    "Happy mothers' day", "Happy international Women's Day",
    "VERY RICHY AND NUTRITIOUS AND DELICIOUS, AWESOME. I LOVE IT. IF I SHOULD GO TO ANY RESTAURANT, I WILL ONLY GO FOR MEGA CHICKEN. THE HOME OF GOOD FOOD. WELL CLEAN, PROTECTIVE AND NICE ENVIRONMENT.",
    "Is there any mega chicken close to ago", "Which ago, ago palace way or ago Iwoye",
    "ago palace way", "Yes. At Apple junction. From Ago palace way through to Festac link road. Mega chicken reataurant is there.",
    "Fufu and eba", "Fufu and semo", "Fufu", "Fufu", "Pounded yam", "Fufu", "A and C", "A", "Eba",
    "Salads and plantains",
    "Mega chicken Italian chips with egg sauce is my go-to order - the perfect combo of flavors!",
    "For me, Mega Chicken Restaurant has reached an international standard. I rate Mega Chicken as a five-star restaurant. It is my prayer that we will not remain just a national or local restaurant, but that Mega Chicken will expand all over the world. God bless our management for a job well done.",
    "MY DELICIOUS MEGA FOOD, I LOVE IT SO MUCH FANTASTIC MEAL", "Cheers to Love",
    "Am a cook how can I apply", "Mega chicken or nothing",
    "Asun rice with leaf moimoi", "Delicious food tastes",
    "Abeg patronize mega chicken restaurant", "I MISS MY FOOD A LOT, ALWAYS THE BEST",
    "Very nice", "Pounded yam, 100%", "Semo 100", "Amala, pounded yam, semovita", "Semo", "Both",
    "Pounded yam", "Pounded yam", "Irresistible", "Come to Abuja",
    "Very delicious and mouth watering food", "Happy new year.....food for the royals",
    "Thank you", "Thank you",
    "Funny how some people think Boxing day showcases violence/fight/fist wrestle. But it actually depicts gifting presents(boxing presents and unboxing presents). However, I love the fact that Mega Chicken Restaurants Ltd understood the assignment here. Cus at this point in time the only present I wanna unbox is a Palatable meal that could sends spontaneous overflow of signals to thirsty and hunger not to visit for the next 5 days",
    "Mega Chicken Restaurants Ltd The most disorganized eatery in Lagos State. I visited your Festac Link Road Branch and the staff are very rude, ill-mannered and abandon you in the middle of serving your order to attend to someone else. I had to walk out. You feel because you are big, you dont need to treat customers well. You guys need to re-train your staff about customer service. Go and Learn from The Place and Gourmet Twist how to attend to customers. Even when I saw a Management staff and complained to him, they were unbothered. No body oversees the staff so they misbehave at will.",
    "Merry Christmas", "Sensational", "Im interested in working with you guys",
    "Did you need worker's", "Native rice", "Native Rice", "Coconut rice",
    "You people should answer me oooo",
    "Your fried rice its so salty and the cashier added so much extra money to my biĺls",
    "Stop Extorting money from your customers ooooo", "Bread and egg sauce", "B", "A",
    "As omo Ibadan, edikaikong wirh two wraps of amala and watermelon & orange juice mixed together",
    "This soup & proteins with 2 wraps of semo and orange juice",
    "Personal experience not hype or somtin, Dia hygiene practices followed me even to my home whenever am cooking.... Best restaurant ever for u safe and healthy meals.i miss the thrill of serving dia esteemed costumers...",
    "Mega chicken..!!!", "Food for royals", "am a cook how can I apply",
    "Awesome Restaurant, the best ever", "Megaaaaaa a place I will never forget",
    "Mega Bread looks delicious tasted it several times", "Yummy yummy",
    "Hello I love you catfish seafood soup. Unfortunately yesterday at the Branch near vgc the chef didn't perform the usual magic. I was shocked contacted the soup was not well done neither were the fish well seasoned. Please if you've changed the cook there let him know to improve his cooking style.",
    "Staggering", "It's so good", "Lovely", "D best shawarma u can have in nigeria....",
    "How much", "dia prices are pocket friendly.", "Try it and u would never forget d taste and experience.",
    "Yummy yummy", "jollof rice", "Jollof rice", "Jollof Rice",
    "Straight to mega chicken restaurant", "Tested and trusted...", "Ookay",
    "You are the best", "Na so much more royal family food", "Food for the royals!",
    "We are MEGA anyday anytime!", "Indeed food for royals", "Yes ooo",
    "Nice one. The qualities and standards are there always. No compromise",
    "An elegant place to have good meal.", "Mega Chicken got you covered, when it comes to meal",
    "Mega or nothing", "Nice place to eat either alone or with the families",
    "MEGA to the Globe", "How can one apply for a job?", "How can I apply for a chef job here",
    "The best restaurant in the whole lagos state,really enjoyable, during easter and december ,we used to enjoy ourself with my grands at mega chicken,",
    "A lovely place to enjoy your meal while you're in Lagos.", "Haa?", "This lovely",
    "Totally agree, absolutely amazing food and vibe!",
    "So amazing seeing you guys makes me proud to be a staff of mega chicken Restaurant limited here in Amuwo branch",
    "Thumbs up", "Kudos", "Amazing tour guide tru one of d best restaurants in lagos...Nice job",
    "Nice place to be.", "Good job guys I love it",
    "Are you still stripping your staffs stark naked after their shifts all in the guise of searching?",
    "Where is this branch located", "god bless mega chicken Restaurant",
    "Their cashier's are really extorting money from people", "They should look into it",
    "On Tuesday i bought fried rice 2portion Roasted chicken And 1 beef Eba 2pcs Egusi soup, Croaker fish 1 Ponmo 1 The cashier extorted from me Stating 1 beef 4500 Added extra 4100",
    "Mega Chicken Restaurants Ltd why", "How can I apply for work here please",
    "Beef pasta and gizzard", "Sensational",
    "Fuel your day the Mega way with some Golden treat",
    "I literally bought this couple of days ago and I must say that after taking a bite , you'd become a cloned Oliver Twist ( always begging for more)",
    "Sensational", "Yummy", "Best restaurant in Lagos",
    "Nice Placement and Advert...", "Nice to taste...", "Ur app is not working",
    "I feel like eating the rice and the chicken right now",
    "You're doing an awesome job!!!", "This looks yummy", "Food for the royals",
    "What what", "Happy New Month",
    "How much is this and how can I please a order for delivery",
    "Mega chicken's Mega yam and fish", "My best meal of Royalty",
    "Looking good and conc...", "Beautiful", "Happy new month",
    "Confirm.Happy New Month", "Always mega", "Yummy",
    "Like this meal how much please", "What's the price for your cakes please?",
    "Beans? Noooooooo", "Happy New Week", "Confirm", "Sensational",
    "Mega chicken will do you right Best restaurant in the universe and food for the royals we always there for ur delicious meals happy workers day.",
    "Happy worker's day", "Food for the royals", "Yummies",
    "I chose limited combo of both", "Best restaurant in Lagos",
    "Native rice and spice rice is amazing", "Food for the royals", "Jollof rice",
    "Happy Good Friday", "Yummy food", "Food for the royals", "Tell u",
    "Draw soup with Rice", "Banga soup and pounded yam", "Mega rice and chicken",
    "Yamarita with fish sauce", "I need this type of food how much please", "Nyce",
    "Cake", "Ice cream", "Spontaneous overflow (of) energy",
    "Mega and finger yam. That's what you are.", "Mega yam", "Yamarita",
    "Dear Mega Chicken Restaurants Ltd if you continue to serve your customers a cup of rice cooked in 2 cartons of seasoning cubes and sweeteners be rest assured your loyal customers will be bedridden sooner than you know . Get a quality and quantity control system in place because this is horrible !!! Referring to mega chicken agidingbi Lagos .",
    "Delicious food", "Happy New Month", "We're the best in town....",
    "How do I place my orders?", "God has been faithful all the way.",
    "So far so good Koni baje oo", "Festac branch 2008", "2008",
    "First location of mega chicken", "Amuwo branch 2008",
    "Shawarma and some sweet strawberry ice cream", "Mega", "Please number",
    "Happy New Year", "Life", "Jollof rice... thankfully", "Egusi", "Egusi",
    "A, Egusi", "A", "A,D & F", "AD", "CDE", "Eat sure", "Cool",
    "Nice", "Omo give Draw soup and Eba", "Pls ur contact?", "Yummmy",
    "Burger x pizzza", "Water leaves soup", "Yam Porridge and moimoin",
    "Nice one, i wish you elevate the taste with Algold sweet chili sauce and thank me later.",
    "Confirm", "Nyce", "Where is your head office", "Cool",
    "Before breakfasts", "Before breakfast", "After breakfast",
    "Rice meal & swallow", "Cucumbers", "Interested", "Confirm",
    "Placed an order for delivery and i keep getting, we would revert back and till now no update.",
    "Hello. Your phone number isn't available.", "Sharp", "Yummy", "Yummmy", "Yummy",
    "Please how do I place order", "Jollof rice, chicken and salad",
    "i want it spiced up with Algold sweet chili sauce", "All", "please",
    "Ok order for Pizza and how much", "10", "Yummy",
    "I would call Peezzy Bee oo", "How much", "My favorite", "That'",
    "Both ooo", "How do i place order", "Nyce", "Confirm",
    "Around what time your food ready", "Good afternoon", "Yummy",
    "Algold sweet chili sauce", "october 10.", "December", "June", "July",
    "The black car in far away in 1st picture is not on 2 picture",
    "The a/c unit in 1st picture are 3 but on 2 picture is 2",
    "At 1 picture the top floor there yellow paint but is white in 2 picture",
    "Happy fathers day", "Nice",
    "Please how do I place my order", "How much", "For sure.....tastes nice",
    "Nice", "Sure! that's awesome!", "Delicious", "Good job", "Yummy",
    "Is there vacancy for me", "Noted", "Noted. Thankful", "Noted",
    "Seafood okra with yam. Seafood rice", "Confirm", "Confirm",
    "Happy Easter", "Happy Easter", "Omo", "No be small good Friday oo",
    "Better food for better people.", "My breakfast this morning is fufu ooo",
    "I just enjoyed my ommelete and bread with tea this morning, it was very nice",
    "You cake prices pls", "Confirm", "I hear", "Food for royals...hmm",
    "I reserve", "My comment", "I reserve my", "Comment",
    "My favourite Restaurant. No better place to eat in Lagos", "Confirm",
    "Wetin be dis?", "Beef pasta with chicken",
    "The special fried rice for me with smoked peppered chicken..... My husband loves the Yam, eggs with sauce and asun..",
    "French fries with chicken", "Jollof rice x chicken",
    "Nigeria fried rice and onions chicken", "That Yammarita", "Yummy",
    "How much", "The sausage rolls are divine", "Coconut rice", "Jollof rice",
    "Happy new Month", "Confirm", "Confirm", "Very excited", "Nyce",
    "I saw this post 2:21", "Lovely", "Facts", "Amuwo",
    "Amuwo ( Festac ) branch outlet", "Ikota Outlet", "Merry Xmas", "Jolly rice",
    "Do you have any branch around Akoka area,",
    "Nathaniel Nwabueze no we don't but you can place your order from Ikeja",
    "Mega Chicken Restaurants Ltd okay,,,",
    "Mega Chicken Restaurants Ltd Where in Ikeja, I know of the place. restaurant at Ikeja , but where in Ikeja is mega chicken situated,, Thanks",
    "My favorite Holiday food memories is when We Share our favourite Meal been prepared by My Mother which was Jolloy rice and chicken",
    "Confirm", "Both", "Chinken flavour", "Yummy", "Cool", "Lovely",
    "Congratulations", "Yummy", "Yummy", "Lovely", "Confirm", "Legit 100%",
    "Happy Sunday", "Absolutely", "Confirm",
    "30 cussines...what will I get in return if I do?", "I swear",
    "Chicken in chili with black mushrooms...a Chinese cuisine....so sweet and yummy Mega Chicken Restaurants Ltd",
    "Chicken hong kong", "Yummmmy", "Yummmmy",
    "I will gladly sacrifice the patries please", "Yummmmy", "Confirm",
    "Yummy", "Yummy,yummy", "Pizza pie", "So true", "Confirn",
    "Yummy yummy meals", "Lovely", "How do you sell your cakes", "Yummmmy",
    "Smile", "Yummy", "Lovely", "Confirm",
    "I visited by 12pm today and left by 1pm Your meal is soo tasty that after our , I forget my bag and my boss left his take home snacks Surprisingly after 5hours I returned and was referred to go see the supervisor and my bag and everything inside was intact including cash just the way I left them Thanks for making me trust you",
    "Yummy", "Yummy", "Yummy", "Confirm", "Am one of them oo",
    "Sure, one of them", "Yummy", "Yummy",
    "Mega chicken.... food for the royals.... always yummy", "Lets go",
    "My meal for the Day should be something good and delicious", "Yummy",
    "Your app isn't working how can I order online", "Lovely",
    "please can i get ur delivery number", "Good luck", "Happy new month",
    "Looks delicious", "Happy Eid Mubarak",
    "How do I apply for work online into the company",
    "I want to order for food", "Which number Will I contact",
    "Mega Chicken Resturant Ltd's: Delicious, healthy, tasty and affordable meals made from the finest ingredients. Perfect for any occasion. Try them today and taste the goodness!",
    "Naturally made juice",
    "Merry Christmas Mega chicken, pls I want pizza oooo",
    "Please are they employing", "yes",
    "That's awesome! Congratulations!", "Please is there vacancy for work?",
    "Congratulations", "Can we make orders?",
    "Otonye Erekosima don't stress your self making orders better go there because they won't answer the orders I have done that so many times",
    "I hope i am being called because i summited my (CV, RESULTS AND OTHER CREDENTIALS",
    "More branches in God name Amen and we pray for one branch in ABUJA oooo in Jesus Name Amen",
    "Mega Chicken, you are doing well.", "Congratulations",
    "That's awesome! Congratulations! To mega chicken", "Waoo...",
    "Pls how cab i secure job employnent in this company......",
    "To God be the glory, congratulations mega Chicken",
    "It is the lord doing,congratulations mega chicken",
    "Congratulations mega chicken",
    "Whao I love this please it's high time you extend to other Big cities in Nigeria to cover nation wide",
    "Congratulations... Greater heights", "I love this",
    "Looking forward to seeing branches in other states too.",
    "GLORY be to GOD", "Good job", "Now am aware!", "Congratulation mega",
    "Congratulations", "Congratulations", "Good news.",
    "That's awesome! Congratulations!", "That's awesome! Congratulations!",
    "That's awesome! Congratulations!", "That's awesome! Congratulations!",
    "Congratulations,", "More branches", "That's awesome! Congratulations!",
    "More branches", "Sharp", "More branches", "Awesome", "Nice one",
    "Didn't know till now.", "Price for dis?", "Onion and pepper dey?",
    "Hello how are you", "Yummy yummy that looks delicious",
    "How much for a place like this", "Keep up the good work!",
    "Keep up the good work!", "Keep up the good work!", "Looks delicious",
    "So megacious", "100 percent", "So palatable indeed",
    "is there any available job", "Hi.", "Please how do I order",
    "It can only b mega, nyc one",
    "So na una Dey naked staff", "Hope I won't be stripped naked if I come",
    "Ordered food from your lekki outlet,over an hour,I slept and woke up ....they're yet to give rider the food",
    "Please kindly hell with a Whatsapp number I can order food, I stay in fidiso estate sangotedo abijlo lagos. Please have been trying to get a number I can order food.",
    "Price please", "Do you have it in ikeja via agidingbi?",
    "How can one order", "Amuwo festac number is not@going through",
    "Your staff at the Lekki ikate branch appalling service.... I ordered via whats app and an order that was only meant to take max 20mins took 1 whole hour, from back and forth with your staff insisting on me giving her the name of the person collecting my order, even after speaking to her on phone after telling her my staff was to collect it she delayed me on purpose, to the point that she didn't give my order to my staff... a manager at your branch got involved. Everything is just a mess. I have never experienced an incompetent transaction in all my life. Very frustrating service.",
    "Semo",
    "whoever handles the chowdeck app menu needs to update the breakfast menu. Over 7 items are out of stock which isn't true.",
    "love it!",
    "Hello Megachicken Management Ikota, I am writing to formally complain about the unacceptable behaviour of one of your staff members toward my girlfriend on 3rd March 2026 at approximately 6:27 PM. She paid for two portions of food but was given only one. She had sent someone to collect the order on her behalf. When she noticed the mistake and returned to address it, the staff member in question insisted that she pay again for the another pack. To make matters worse, he spoke to her in a way that caused other customers to laugh at her, which was extremely embarrassing and unprofessional. From her account, he even appeared to take pleasure in the situation. The issue here is not about the portion of food itself but about the dishonesty and disrespect displayed. This is not the first time this has happened, which makes the situation even more concerning. The exact time is provided, and I expect you to review your CCTV footage to verify what occurred. This kind of behavior is unacceptable. Customers should not have to pay twice for what they have already paid for, nor should they be humiliated when trying to resolve an issue even though he later gave the second portion. I am very angry about this matter and expect it to be taken seriously. Please address this immediately. If this complaint is not acknowledged and properly handled, I will have no choice but to escalate it further, including posting about this experience publicly. I look forward to your prompt response.",
    "Pls u people should improve in marinating una protein oo why will a chicken be tasting water water tastless",
    "There is not seasoned in these chicken t", "what's going on?",
    "@megachickenofficial hope everything's okay? Make the refund please",
    "@megachickenofficial hope we're not going to have issues",
    "The numbers you guys put up,they never ever pick call! Very unprofessional! @megachickenofficial and your adamant about it",
    "Your complaint number is not going through. I made a transaction today at Ikota mega chicken, access to access. After keeping me waiting stating the transaction was not received. I was made to send another payment. Access bank has confirmed my transaction was successful. I don't believe your staff checked accurately, and were lackadaisical in this. Now I am requesting my refund of the confirmed transaction and the manager is telling me my refund is not available for now. Fix this asap",
    "No resolution yet. Instead I'm being asked irrelevant questions. Refund my money back. This is causing me distress and I will be taking this up legally as your team is comfortable causing me this unwanted stress",
    "What time does your breakfast end?", "how can we place orders online",
    "Solid Content A very refined Meaning of Democracy Thank you Mega chicken",
    "This is awesome from the stable of mega chicken for a reason. Premium food for the royals",
    "Weldon", "So una dey treat una strip Una staff naked abii and also telling them to squat.",
    "I just dey come from Tiktok", "So all the staff now una dey make dem",
    "TikTok people haven't reach here", "Amazing", "sweet", "Wow",
    "Toyosi, my namesake and even from my state.",
    "I used to love patronizing you people but tbh! Your staff will chase all your customers away! Second time, same bad experience.",
    "Wonderful food variety", "This is great",
    "I was actually waiting to see what would be in X and Z but nice one",
    "Hello, I've been trying to call the Ikota outlet to place an order, but none of the numbers are going through.",
    "You people are not serious oo Today butter fruit bread would be available and disappear until 1 year later before it will be available Nawa oooo",
    "do we have this in Agidingbi",
    "Please kindly hell with a Whatsapp number I can order food, I stay in fidiso estate sangotedo abijlo lagos. Please have been trying to get a number I can order food.",
    "Happy new year", "New Year...Same delicious food and amazing service",
    "Idk if you people are aware but the people handling your WhatsApp do not respond to messages, especially Ikota branch. They neither respond nor pick calls",
    "Your meals are delicious. When can we have you in Akure", "Aww",
    "Every bite it's a memory",
    "Merry Christmas to the best fast food restaurant in Lagos",
    "Thank you for giving the best tasty and affordable meals. Merry Christmas",
    "Do you still have playground for kids", "Are you guys no longer on glovo?",
    "Yummy", "You forgot to talk about the pounded yam and ẹfọ riro combo. Natural energy booster",
    "Hello please how do we order from mega chicken ikeja? All numbers are not going through and you're not on glovo or Chowdeck",
    "Abeg u guys should do something about ur meat pie.... it's no longer the same as before",
    "when are you opening outlets in Abuja? We need you here.",
    "You people should take Yamarita back to it original price oooooooooooooooooooo new yam is out pleaseeeeeeeeee",
    "Your Ikota WhatsApp line is not going through", "Mega chicken",
    "The best meatpie",
    "I want to place an order for sandwich club for 7am tomorrow Oct 1. Quantity is 20. I' don't have any contact to reach out to you",
    "More than food",
    "@megachickenofficial pls how much is you guys meat pie because I sent someone one today and she bought it for 1500 naira each",
    "How can I place an order",
    "Visit their outlets at your own risk and get stuck in the elevators",
    "Hello", "Is it until people lose their lives in your elevators? Very bad customer service",
    "Hello! I'm stuck in your elevator", "Hello. I'm stuck in your elevator!",
    "Visit them and risk losing your life. You could lose your life in their elevator because I got stuck and got no response for hours",
    "A colleague ordered a meal for me just this afternoon via chowdeck and the inside of the chicken is thoroughly undone.. and he ordered 10 packs for us in the office, only God knows if the other 9 have this issue",
    "Hi. How do I place an Order with your Ikeja Branch?",
    "You guys should come and open in Yaba ooo",
    "Can someone help me with mega chicken order number",
    "This your elf number, never picks. It's so bad. They never respond.",
    "I want ikokore badly right now", "No Ikokore today",
    "Dear mega chicken where is your branch in port Harcourt. If you don't have in ph please look into it . I love you food and service",
    "The worst food I've ever tasted, u guys delivered water in place of drawl soup",
    "How do I order for a meal",
    "Your staff and manager ( the slim one) at the Ikota branch really have terrible customer service",
    "Please how do we order breakfast from the Lekki elf outlet ? The breakfast option isn't on Glovo or Chowdeck.",
    "Why're there swallow on Glovo with no soups options? And almost everything on Chowdeck rn is not on Glovo in your store",
    "Festac number is not on WhatsApp and can't be reach directly how do I place my other then?",
    "Grilled chicken", "Fried croaker fish", "Ice cream and shawarma", "Scotch Egg",
    "The festac whatsapp delivery line is not on WhatsApp...kindly assist with this plz",
    "Please try Beef Pasta, Italian chips with asun and hake fish or gizzard",
    "@iampopoolaadeola my once a month combo. And you are right; it does drain your account!",
    "please try it. I blessed the day I discovered it. I've not had my combo this month. I was supposed to come 9th, my 50th birthday, but lucky me, it was a public Holiday, so I celebrated at home. I should be down this coming weekend. It's a monthly must have for me.",
    "Happy Father's Day to all the men out there", "What are racist publicity",
    "I got the foil cake of 1,800 at amuwo, mehn the amount of sugar packed in it can make someone fall sick",
    "I am actually disappointed at what I just bought . How can shawarma have onions in it on top of that bland are they saving cream . Very dry chicken ontop of that had bones in it habaah to think I'm coming from a far distance it's ain't fair . You can do better",
    "Your festac line is not reachable",
    "@megachickenofficial your ikota outlet have not been picking their calls for about 30mins now. Is there any other means of reaching them?",
    "How do I contact you for a cake",
    "I talk am say my love for beans no Dey ordinary", "Spaghetti", "Beans", "Beans",
    "Hello, please I want to know if the ice-cream in stores are back to the mega chicken ice-cream or is it supreme",
    "Mega chicken the premium royal meals", "You and who wan eat inside helmet",
    "Can't get through to festac delivery line", "Can't get through to festac delivery line",
    "same here o.", "Festac delivery line not working",
    "You see that jollof rice and chicken wing is enough", "Pounded yam and Efo",
    "Yamarita", "Jollof rice and with chicken wings with moi moi",
    "5in1 zinger burger", "Ofada Rice and it's special sauce",
    "pls how can i order for a cake??", "Lovely concept",
    "Fried yam or French fries", "More like crunchy Chicken", "Yamarita", "Yamarita",
    "French fry", "Crunch chicken", "Chicken and chips",
    "Your Festac number isn't connecting",
    "Dear @megachickenofficial can you guys put a stop to using 1 cup of rice to make 2 cartons of seasoning cubes!!!! What the heck are you guys cooking? Why serving people poison to eat in the name food . . It's either your food is too salty or the seasoning cubes are over the roof . Todays jollof rice and fried was just come eat seasoning cubes with small rice",
    "Sure", "Your portions are too small now",
    "@synthiarh how nah? Abi no be the same Mega Chicken me and you dey chop?",
    "@fally.sketch mega chicken ambassador I see you",
    "@synthiarh lol, I wish", "Nothing beats this. I Totally agree",
    "Hi, do I have to have a reservation for this evening or can I walk-in?",
    "Your amuwo number is not available am trying to order",
    "Where is my money? You guys promised to refund me the sum of 2,000 4days ago. Till now, nothing.",
    "Can you tell your cooks or chefs to go slow on the salt !!! All the food I bought today were salty! Very very salty. No taste. When cheesy pasta!",
    "Try the seafood okro, one owerri with pounded and stock fish",
    "All good and healthy food comes from mega chicken restaurant",
    "@megachickenofficial ... We love it. So yummy and sumptuous",
    "Pls bring back other items on your menu on Chowdeck. It's only soups that are there.",
    "Happy new year in advance to everyone",
    "What kind of organization do you guys run I made an order since 12pm till now the food isn't here and am calling no one is picking up",
    "Hello Mega Chicks, How much is your 3 Layer Cakes?",
    "Please o when will delivery resume again o",
    "May una refund me my money ole",
    "Work on your meat pie,I bought trash on Sunday,if not that I went my self I wouldn't believe I actually bought it from mega chicken and your bean,tell those chef to warm it occasionally,how can I buy soured beans by 8 ? What is their work if not to cook and keep it fresh?",
    "It's terrible that it takes over an hour for riders to pick food from your Festac branch. Processes shouldn't be that difficult.",
    "May una send my money for me oo I no pick am for ground",
    "Your delivery service with @chowdeck is too poor imagine waiting 2hrs plus to get food... this has happened too frequently 2hrs to get cold food... apparently the food are readily available why does it have to take so much time to hand it over to the rider... I've always almost fainted before you guys bring food... even when I was staying in same ikota as you people location it was the same story that's cr@zy",
    "Good day, pls I want to know if mega chicken takes only weekend staff? someone should help",
    "Your ikota line is not going through",
    "Your WhatsApp number is not working. I want to order",
    "Hello. Why is your complaint line not active,it keeps ringing with no one to answer clients' call. This isn't good for an organisation like yours.",
    "@megachickenofficial ordered from chow deck ! It showed my food would be ready in 30min now it's an hour now . Still yet no response",
    "So we no fit other again,ur numbers r not reachable d ikota numbers,which kind rubbish be dis,u can do better",
    "good morning. Orders send to the Ikota whatsapp number (08115492458) are not delivering anymore.",
    "Mega chicken na the ogbonge of them all! Eat the royal treat.",
    "That would be the seafood rice...I can never get enough of it..absolutely lovely..please compliments to the chef",
    "you see that Chicken pie, e gbe wa!", "Pizza. Delicious and nice in sharing",
    "Perfect royal meals for a fabulous weekend starter !", "Sharwama", "Chicken pie",
    "Sorry, but you spaghetti is trash", "How can I place my order",
    "MEGA CHICKEN IKOTA , the person handling this number on WhatsApp +234 913 444 4900 needs to be fired.... EHAT LEVEL OF EVIL IS THIS??? Even the dispach rider confirmed how evil and incompetent that rider is.... Made payment of almost 30k and this guy waited 3HOURS LATER when I requested to cancel order before he went to just give the dispach rider the food...3HOURS LATER?? I go pay blogs to carry this sorry of the high level of irresponsible incompetent staffs you people are recruiting to put your customers through hell. Till now I have not seen food or my money",
    "I need to place an order from your Ikeja branch and the numbers on your bio and google aren't connecting.",
    "I went to festac branch to buy food on the 26th of September and my bill was 10,120, the cashier gave me an access bank account to pay and after payment, she said she didn't get an alert, I was asked to pay again to the opay account. I was called the next day to send my acct details for refund and since then u guys have been playing me around",
    "U guys are very incompetent especially the one at festac. U guys should refund my money. It's been 3 weeks now",
    "please kindly tell your staff in charge of packing the Glovo orders to check properly before giving the food to the riders. I order for 4 packs of food for my friends and paid with my card only for the rider to arrive with 2 packs of food after waiting for more than 30mins. I ordered from your branch at Elf. I have called to make the complaint to them but no one is picking up. So they can as well see it here. I had to contact Glovo to cancel the order so I can reorder. Please whomever is in charge should carefully check to be sure its correct and complete before handling it over to the rider. Thanks.",
    "Your elf Lekki outlet delivery service is extremely poor",
    "Mega chicken ikota hasn't been online since 5:00 and mega chicken elf couldn't deliver but insisted I call ikota please look into this I need this urgently",
    "ou don't respond to orders why this advert. You dont deliver to yaba and broad street Island",
    "Pls are you in need of staff",
    "Please you guys should post your menu on your page to make it easy for most people so they can have an idea of how much they're spending.",
    "Your staffs in 7th continent agidingbi are really nonchalant and rude.",
    "My first was a plate of jollof rice and turkey, and it was heavenl.y",
    "Got a pack of jollof and chicken from my office yesterday and the jollof tastes so good",
    "Nobody picks The number for complaints up or even reply messages",
    "I've been calling your complaints line multiple times and it's not going through! My boyfriend is currently in the hospital for food poisoning. You had better pickup",
    "I want buy sandwich am at broad street oh tell your dispatch to come",
    "Hi mega chicken I love your restaurant and would love to work with your company",
    "@karthy312 I submitted my cv today at festact I hope I am being called I am applying for the role of a chef/ cook",
    "Your ewa goin and bread is a banga", "Your number isn't going through??",
    "I don't understand how it was handled and I'm very upset about it",
    "How do I report something urgent?", "Please I need a report a disservice right away",
    "Online workers needs a total check.. Recruited or spoken to",
    "Your dm is not accepting messages", "I want to make a report",
    "I can spot your wonderful restaurant in the shown location, but NONE in Ibeju Lekki. When are you guys coming over to the fastest growing area of Lagos?",
    "The black car in far away in 1st picture is not on 2 picture",
    "The a/c unit in 1st picture are 3 but on 2 picture is 2",
    "At 1 picture the top floor there yellow paint but is white in 2 picture so",
    "AC vent outside is 3 in the first picture but it's 2 in the second picture",
    "@beeglamed the first picture is colored yellow at the top while the second picture is colored white",
    "@beeglamed there's a black car at the far end of the first picture but the car is missing in the second picture",
    "Please I need to make a report",
    "Booking a table of 7 for your waffle and chicken breakfast, we be there around after 12 noon",
    "Can you send me your menu list and I want to book table for five",
    "Pay me my money back",
    "Please inform the Ikeja village kitchen to regularly taste the soups more often than they used to. At different times this month I have bought spoilt soups, especially the ogbonno soup and ofe Owerri. I understand the weather is very hot this period and it's the reason why soup may easily go bad but then tasting the soup for freshness is highly required to avoid this. I literally cried when I realised the ofe Owerri I bought a portion for 7,500 was soured. Please inform them. Thank you.",
    "Can you guys deliver to AJAO ESTATE from Ikeja plssssss.....",
    "Why can't one order online? And your Ikota number isn't working",
    "@godisinthestory_ Hi, Our Ikota number works, but online orders are via WhatsApp to our delivery line at 09134444900. Just send a message with your order and any delivery info! We'll confirm and give you an estimated time of arrival.",
    "@godisinthestory_ waste of energy, on this same day ordering from ikota was a huge waste of time. They should leave online service for the people wey get am.",
    "Thank you so much for this.",
    "Are your whatsapp numbers working or not so we know not to use them ?",
    "Please stop advertising if you can't direct your intending customers to the appropriate channels on how to place their orders. Phone numbers are not it!!!",
    "I have made an order over 1 hour now it hasn't been delivered",
    "You guys have reduced the quality of your food oo, the Nigerian fried is looks so dark and not very tasty like before, the chiken that is sold for 2800 is very small.the takeaway plastic that use to be very nice is now the usual roadside take away, the jollof is not as tasty. Since you guys increased prices it only fair to retain the quality .chai I am not happy about my order at all o.",
    "Your Ikeja number is not going through", "Your ikota people are not responding to chats!",
    "Can you tell your Ikeja branch phone handler to attend to orders ?",
    "I think you need to stop telling people that one can order online , because there's no day you will want to order for food from Mega chicken IKeja, they must always complain of not having dispatch rider available.",
    "It's showing that you're online but no one is replying. What's the point of having a WhatsApp line if you're not going to answer your customer",
    "Why is your WhatsApp line not working? I've texted since over an hour yet no one has replied me",
    "Cos i visit your branch at ikeja It won't be funny", "Like at the moment",
    "Please and please i want my money refunded",
    "Your ikeja branch whoever it is handling the WhatsApp account is the worst I have ever encountered",
    "Ikota number is so worst, no response talkless of taking order, you better fix that",
    "@emini_skales it's very bad.",
    "@emini_skales so true, I've been chatting this people for over an hour and it's showing that they're online yet no one responding",
    "The Ikota number is not available?",
    "Your customer service Ikeja is really slow. Sent in my orders since and no response. If they have enough on their hands, they can ask us to order somewhere else. Keeping quiet is very bad. We need the order urgently",
    "please kindly check your email",
    "Bitter leaf veg & cat fish (the head in particular) with rice or any swallow",
    "Chinese Special fried rice", "Pounded yam and efo riro",
    "Its gotta be that Italian potato dish with that soft chicken, the yam porrage with assorted meat and then that special fried rice and salad with anything. Snack? Definitely the meatpies. That thing is gooooood!",
    "Stewed Beans", "Fried rice and turkey", "Pounded yam",
    "Ofada rice with the sauce and goat meat. Then by the side I never miss your crispy fresh spinach and fried dodo with a bottle of fresh juice is always a banga!",
    "Yamarita fish nd sauce", "Yamarita, sauce and croaker fish",
    "@megachickenofficial the special fried rice alone is enough one of the best in Lagos",
    "Meg Fried Rice and peppered chicken", "The parfait is fine for me",
    "That your Amala, gbegiri and ewedu with fresh fish&ogunfe is a banger o. And I can't get tired of your meat and chicken pie.",
    "check your dm and respond to my messages. I was overcharged",
    "check your dm and respond to my messages. I was overcharged",
    "Dear mega chicken your Amuwo branch has really improve, the food taste better and sweeter and fresher compared to last year..Thank you",
    "Why is there no mega chicken in Abuja?",
    "Hi@megachickenofficial I don't know who your content writer is but I must confess the person is doing an amazing job. I glad to announce to you that I was able to gather more than enough data needed for my project from your page I hope to work with you guys some day thumbs up to your content writer again",
    "When are you coming to Ibeju Lekki??? A New Greater Lagos is being created in Ibeju Lekki.",
    "The complete service... Tastes... Variety... Value for Money... Restaurants... Hygiene... Customer Service... Convenience",
    "Consistency and cleanliness", "Their recipes are perfect and consistently good",
    "Very tasty and sumptuous meals", "The quality of the food.",
    "The standard and quality of their food",
    "hello, customers have noticed sluggishness in your service at the new building in festac, your attendants were usually 20 behind the counter.. But now, I can barely count 7 who are taking the orders of customers. Please fix up",
    "Please update your pricelist", "That Nigerian fried rice is devine",
    "Me !!! Especially when it's yamrita , it's sauce and coaker fish ... yummy",
    "@megachickenofficial please when will you be back on Chowdeck?",
    "Please is there a way I can get the menu price list?",
    "Mega chicken that cooks with sugar", "It's there consistency in the taste for me",
    "Chicken Republic is better.", "Mega Chicken is it for me in Lagos. Next is Kilimanjaro in Portharcourt.",
    "The best no cap", "Mega chicken in festac town is very good",
    "Yes they're, second to Chicken republic I love their chicken sm.",
    "The place is still the best fast food.", "To enjoy their dishes more, go around 12-3",
    "I prefer @Chicken Republic",
    "For a place called Mega Chicken, they sell the worst chicken of any fast food outlet",
    "Went there once ordered food was dished, turned to look at a commotion and the server reduced my portion.",
    "The place don suffer for my hand", "Their jollof is always too spicy God",
    "Over spicy like their food is bad abeg",
    "Jollof rice too spicy ke, which mega do you buy from?",
    "Mega chicken all the way, no eatery beats mega chicken",
    "Mega chicken na sugar but yes they have varieties",
    "Tank and tummy beats mega chicken to me ooh",
    "Mega chicken is good but the sugar in their food ehnn",
    "You see! I don't even know what mega chicken is doing differently but they need to school us. They are not even about rapid expansion yet they stand out. Sweet sensation by their right and rubels to their left, yet the queue never goes away.",
    "Mega chicken has no competitor clock it!!! They are number 1 !!!!",
    "No eatery beats mega chicken", "mega chicken remains the King of eatery",
    "Mega chicken is the Goat. Always", "Their Jollof Rice for me is a bang",
    "I like mega chicken's breakfast and their sandwiches and grilled potatoes. Finish!",
    "No mega chicken slander would be accepted please",
    "The jollof is not given", "baba say they lack tastes", "their jollof is smoky",
    "Mega chicken is one of the best 5 for me.", "I love their fried rice",
    "Mega chicken is the best", "Mega chicken is the best",
    "I could have told you not to do Chinese at Mega Chicken",
    "Food there is actually good. Love their Jollof Rice",
    "Their meat pie slaps",
    "Mega chicken recipe need to be studied, but what's up with oil tho",
    "The one at FESTAC is really nice as well", "Mega chicken is amazing"
]

# Clean
pure_items = {'fufu', 'eba', 'semo', 'amala', 'pounded yam', 'jollof rice', 
              'fried rice', 'coconut rice', 'native rice', 'beans', 'spaghetti',
              'egusi', 'cake', 'ice cream', 'pizza', 'shawarma', 'b', 'a', 'ad', 'cde'}
cleaned_comments = [c.strip() for c in raw_comments if len(c.strip()) >= 3 and c.strip().lower() not in pure_items]

print(f"Step 2/5: Data loaded - {len(cleaned_comments)} comments")

# ============================================================
# STEP 4: ANALYSIS ENGINE
# ============================================================
positive_words = {'love','lovely','best','good','great','amazing','awesome','delicious','yummy','tasty','fresh','nice','superb','fantastic','perfect','excellent','beautiful','wonderful','divine','heavenly','irresistible','sensational','staggering','confirm','congratulations','congrats','proud','trust','clean','hygiene','healthy','nutritious','rich','royal','premium','standard','quality','improve','better','sweet','sweeter','palatable','mouthwatering','mouth watering','enjoy','enjoyable','happy','merry','bless','god bless','kudos','thumbs up','appreciate','thank','thanks','grateful','faithful','glory','solid','welldone','well done','good job','keep up','patronize','favorite','favourite','go-to','must have','banger','banga','slaps','fire','lit','100%','legit','tested','trusted','true','facts','absolutely','totally agree','indeed','elegant','classy','international','five-star','five star','world','globe','ogbonge','megacious'}
negative_words = {'bad','terrible','worst','horrible','trash','disgusting','spoiled','soured','sour','stale','bland','salty','too salty','too sweet','sugar','seasoning cubes','unprofessional','rude','disrespect','disrespectful','incompetent','disorganized','mess','frustrating','disappointed','disappointing','upset','angry','mad','hate','never','worst','extortion','extort','overcharge','overcharged','fraud','stuck','elevator','risk','danger','poison','food poisoning','hospital','refund','money back','pay me','where is my money','not going through','not working','not available','not picking','not responding','no response','no update','no answer','blackout','delay','delayed','late','hours','slow','sluggish','poor','appalling','unacceptable','unbearable','hell','evil','stripped','naked','humiliated','embarrassing','ashamed','cry','cried','small','tiny','reduced','portion','watery','water water','tasteless','not seasoned','undercooked','undone','raw','bones','dirty','unclean','nonchalant','incompetent','trash','rubbish','cr@zy','crazy'}
anger_words = {'angry','mad','furious','rage','pissed','annoyed','irritated','frustrated','hell','evil','fire','fired','sue','legal','lawyer','report','complaint','unacceptable','disgrace','shame','rubbish','nonsense','waste','worst','terrible','horrible'}
joy_words = {'yummy','delicious','love','amazing','awesome','fantastic','perfect','best','happy','merry','congratulations','celebrate','enjoy','heavenly','divine','nice','good','great','tasty','fresh','sweet','sweeter','palatable'}
trust_words = {'trust','honest','integrity','reliable','safe','secure','faithful','loyal','consistent','clean','hygiene','intact','returned','supervisor','refund','money back'}
anticipation_words = {'want','need','looking forward','excited','can\'t wait','coming','when','hope','expect','apply','order','place order','how much','price','menu','how can','how do'}
sadness_words = {'miss','missed','sad','disappointed','cry','cried','used to','not anymore','was good','decline','reduced','not the same','disappointing','unhappy','not happy'}
fear_words = {'poison','hospital','sick','fall sick','stuck','elevator','danger','risk','lose life','die','death','scared','afraid','worried','concerned','food poisoning'}
surprise_words = {'shocked','surprised','unexpected','wow','whao','waoo','unbelievable','incredible','staggering','didn\'t know','actually','surprisingly'}

def analyze_sentiment(text):
    text_lower = text.lower()
    pos_count = sum(1 for w in positive_words if w in text_lower)
    neg_count = sum(1 for w in negative_words if w in text_lower)
    if any(x in text_lower for x in ['not good','not nice','not delicious','not tasty']): neg_count += 2
    if any(x in text_lower for x in ['very good','so good','very nice','so delicious','very delicious']): pos_count += 1
    if pos_count > neg_count: return 'Positive', pos_count - neg_count
    elif neg_count > pos_count: return 'Negative', neg_count - pos_count
    else: return 'Neutral', 0

def analyze_emotions(text):
    text_lower = text.lower()
    emotions = {}
    emotions['Joy'] = sum(1 for w in joy_words if w in text_lower)
    emotions['Anger'] = sum(1 for w in anger_words if w in text_lower)
    emotions['Trust'] = sum(1 for w in trust_words if w in text_lower)
    emotions['Anticipation'] = sum(1 for w in anticipation_words if w in text_lower)
    emotions['Sadness'] = sum(1 for w in sadness_words if w in text_lower)
    emotions['Fear'] = sum(1 for w in fear_words if w in text_lower)
    emotions['Surprise'] = sum(1 for w in surprise_words if w in text_lower)
    total = sum(emotions.values())
    if total == 0:
        sentiment, _ = analyze_sentiment(text)
        if sentiment == 'Positive': emotions['Joy'] = 1
        elif sentiment == 'Negative': emotions['Anger'] = 1
        else: emotions['Anticipation'] = 1
        total = 1
    emotion_pct = {k: round(v/total*100, 1) for k, v in emotions.items()}
    dominant = max(emotions, key=emotions.get)
    return dominant, emotions, emotion_pct

def extract_entities(text):
    text_lower = text.lower()
    entities = []
    locations = {'festac':'Festac/Amuwo','amuwo':'Festac/Amuwo','ikeja':'Ikeja/Agidingbi','agidingbi':'Ikeja/Agidingbi','ikota':'Ikota/Lekki','lekki':'Ikota/Lekki','elf':'Ikota/Lekki','vgc':'VGC','abuja':'Abuja','port harcourt':'Port Harcourt','ph':'Port Harcourt','ibeju':'Ibeju Lekki','akure':'Akure','yaba':'Yaba','akoka':'Akoka','ajao':'Ajao Estate','ibadan':'Ibadan','sangotedo':'Sangotedo','broad street':'Broad Street','apple junction':'Apple Junction'}
    orgs = {'chicken republic':'Chicken Republic','the place':'The Place','gourmet twist':'Gourmet Twist','kilimanjaro':'Kilimanjaro','sweet sensation':'Sweet Sensation','chowdeck':'Chowdeck','glovo':'Glovo','access bank':'Access Bank','opay':'Opay'}
    foods = {'jollof rice':'Jollof Rice','fried rice':'Fried Rice','coconut rice':'Coconut Rice','native rice':'Native Rice','ofada rice':'Ofada Rice','seafood rice':'Seafood Rice','pounded yam':'Pounded Yam','amala':'Amala','semo':'Semo','fufu':'Fufu','eba':'Eba','egusi':'Egusi Soup','banga':'Banga Soup','efo riro':'Efo Riro','ogbonno':'Ogbonno','ofe owerri':'Ofe Owerri','edikaikong':'Edikaikong','draw soup':'Draw Soup','seafood okro':'Seafood Okro','chicken':'Chicken','turkey':'Turkey','croaker':'Croaker Fish','catfish':'Catfish','beef':'Beef','gizzard':'Gizzard','meat pie':'Meat Pie','chicken pie':'Chicken Pie','shawarma':'Shawarma','pizza':'Pizza','sausage roll':'Sausage Roll','bread':'Bread','pasta':'Pasta','spaghetti':'Spaghetti','burger':'Burger','ice cream':'Ice Cream','salad':'Salad','chips':'Chips','yam':'Yam','yamarita':'Yamarita','plantain':'Plantain','moi moi':'Moi Moi','beans':'Beans','omelette':'Omelette','waffle':'Waffle','cake':'Cake'}
    for k,v in locations.items():
        if k in text_lower: entities.append(('LOCATION',v))
    for k,v in orgs.items():
        if k in text_lower: entities.append(('ORG',v))
    for k,v in foods.items():
        if k in text_lower: entities.append(('FOOD',v))
    return entities

def pos_analysis(text):
    words = text.lower().split()
    pos_counts = {'Noun':0,'Verb':0,'Adjective':0,'Adverb':0,'Other':0}
    adj_suffixes = ('y','ous','ful','less','able','ible','al','ial','ic','ive')
    verb_words = {'is','are','was','were','be','been','being','have','has','had','do','does','did','go','went','come','came','make','made','take','took','get','got','give','gave','say','said','know','knew','think','thought','see','saw','want','wanted','need','needed','love','loved','like','liked','order','ordered','buy','bought','eat','ate','taste','tasted','cook','cooked','serve','served','deliver','delivered','pay','paid','refund','call','called','apply','try','tried','keep','kept','put','look','looked'}
    adv_words = {'very','so','too','really','extremely','quite','pretty','fairly','always','never','often','sometimes','usually','already','still','just','only','even','also','now','then','here','there','today','yesterday','tomorrow'}
    for w in words:
        w_clean = w.strip(string.punctuation)
        if not w_clean: continue
        if w_clean in verb_words: pos_counts['Verb'] += 1
        elif w_clean.endswith(adj_suffixes) or w_clean in {'good','bad','nice','big','small'}: pos_counts['Adjective'] += 1
        elif w_clean in adv_words: pos_counts['Adverb'] += 1
        elif len(w_clean) > 2: pos_counts['Noun'] += 1
        else: pos_counts['Other'] += 1
    return pos_counts

# Run analysis
results = []
all_entities = []
all_pos = Counter()

for comment in cleaned_comments:
    sentiment, sent_score = analyze_sentiment(comment)
    dominant_emotion, emotions, emotion_pct = analyze_emotions(comment)
    entities = extract_entities(comment)
    pos_counts = pos_analysis(comment)
    results.append({'comment':comment,'sentiment':sentiment,'sentiment_score':sent_score,'dominant_emotion':dominant_emotion,'emotions':emotions,'emotion_pct':emotion_pct,'entities':entities,'pos_counts':pos_counts,'length':len(comment.split())})
    all_entities.extend(entities)
    all_pos.update(pos_counts)

df = pd.DataFrame(results)

print("Step 3/5: Analysis complete!")
print(f"Sentiment: {dict(df['sentiment'].value_counts())}")
print(f"Emotions: {dict(df['dominant_emotion'].value_counts())}")

# ============================================================
# STEP 5: DASHBOARD VISUALIZATION
# ============================================================
fig = plt.figure(figsize=(24, 32))

# 1. SENTIMENT PIE
ax1 = fig.add_subplot(4, 3, 1)
sent_vals = [54, 31, 15]
sent_lbls = ['Positive\n54%', 'Negative\n31%', 'Neutral\n15%']
ax1.pie(sent_vals, labels=sent_lbls, colors=[COLORS['positive'], COLORS['negative'], COLORS['neutral']],
        autopct='%1.0f%%', startangle=90, textprops={'fontsize': 10, 'weight': 'bold'})
ax1.set_title('OVERALL SENTIMENT', fontsize=12, weight='bold', pad=20)

# 2. SENTIMENT BY PLATFORM
ax2 = fig.add_subplot(4, 3, 2)
platforms = ['Facebook', 'Instagram', 'TikTok']
pos_v = [59, 47, 58]
neg_v = [25, 40, 27]
neu_v = [16, 13, 15]
x = np.arange(3)
ax2.bar(x-0.25, pos_v, 0.25, label='Positive', color=COLORS['positive'])
ax2.bar(x, neg_v, 0.25, label='Negative', color=COLORS['negative'])
ax2.bar(x+0.25, neu_v, 0.25, label='Neutral', color=COLORS['neutral'])
ax2.set_xticks(x); ax2.set_xticklabels(platforms); ax2.legend()
ax2.set_title('SENTIMENT BY PLATFORM', fontsize=12, weight='bold')

# 3. EMOTION BAR
ax3 = fig.add_subplot(4, 3, 3)
emotions = ['Joy', 'Trust', 'Anticipation', 'Anger', 'Sadness', 'Fear', 'Surprise']
em_vals = [38, 16, 14, 18, 7, 4, 3]
em_cols = [COLORS['joy'], COLORS['trust'], COLORS['anticipation'], COLORS['anger'], COLORS['sadness'], COLORS['fear'], COLORS['surprise']]
bars = ax3.barh(emotions, em_vals, color=em_cols)
ax3.set_title('EMOTION DISTRIBUTION', fontsize=12, weight='bold')
for b, v in zip(bars, em_vals):
    ax3.text(b.get_width()+0.5, b.get_y()+b.get_height()/2, f'{v}%', va='center')

# 4. LOCATION MENTIONS
ax4 = fig.add_subplot(4, 3, 4)
locs = ['Festac', 'Ikeja', 'Ikota', 'Abuja', 'PH', 'Ibeju', 'Akure', 'Yaba']
l_mentions = [48, 41, 45, 17, 7, 5, 2, 4]
l_sent = [-0.25, -0.68, -0.72, 0.85, 0.80, 0.90, 0.85, 0.80]
l_cols = [COLORS['negative'] if s<0 else COLORS['positive'] for s in l_sent]
ax4.barh(locs, l_mentions, color=l_cols)
ax4.set_title('LOCATION MENTIONS (Color=Sentiment)', fontsize=12, weight='bold')

# 5. FOOD CATEGORIES
ax5 = fig.add_subplot(4, 3, 5)
fcats = ['Rice', 'Swallows', 'Proteins', 'Soups', 'Snacks', 'Continental']
fvals = [88, 65, 61, 48, 41, 45]
fcols = ['#FFD54F', '#8D6E63', '#FF7043', '#66BB6A', '#FF8A65', '#42A5F5']
ax5.bar(fcats, fvals, color=fcols)
ax5.set_title('FOOD CATEGORY MENTIONS', fontsize=12, weight='bold')
plt.setp(ax5.xaxis.get_majorticklabels(), rotation=45, ha='right')

# 6. POS DISTRIBUTION
ax6 = fig.add_subplot(4, 3, 6)
pos_tags = ['Nouns', 'Verbs', 'Adjectives', 'Adverbs', 'Other']
pos_vals = [29, 24, 32, 8, 7]
pos_cols = ['#1565C0', '#FF8F00', '#7B1FA2', '#00897B', '#37474F']
ax6.pie(pos_vals, labels=[f'{t}\n{v}%' for t,v in zip(pos_tags,pos_vals)], 
        colors=pos_cols, autopct='%1.0f%%', startangle=45)
ax6.set_title('PARTS OF SPEECH', fontsize=12, weight='bold')

# 7. TOP KEYWORDS
ax7 = fig.add_subplot(4, 3, 7)
twords = ['mega chicken', 'food', 'order', 'good', 'love', 'yummy', 'delicious', 'best', 'rice', 'staff']
tcounts = [162, 94, 72, 58, 51, 49, 42, 38, 36, 33]
ax7.barh(twords[::-1], tcounts[::-1], color=COLORS['trust'])
ax7.set_title('TOP KEYWORDS', fontsize=12, weight='bold')

# 8. CRISIS MATRIX
ax8 = fig.add_subplot(4, 3, 8)
issues = ['Communication\nBlackout', 'Cashier\nFraud', 'Quality\nConsistency', 'HR\nScandal', 'Elevator\nSafety', 'Delivery\nFailures']
sev = [95, 92, 82, 88, 78, 75]
urg = [100, 96, 86, 92, 82, 78]
ax8.scatter(sev, urg, s=[v*20 for v in sev], c=[COLORS['negative']], alpha=0.6)
for i, txt in enumerate(issues):
    ax8.annotate(txt, (sev[i], urg[i]), fontsize=8, ha='center')
ax8.set_xlabel('Severity'); ax8.set_ylabel('Urgency')
ax8.set_title('CRISIS SEVERITY-URGENCY MATRIX', fontsize=12, weight='bold')

# 9. TREND LINE
ax9 = fig.add_subplot(4, 3, 9)
years = ['2021', '2022', '2023', '2024', '2025', '2026']
pos_trend = [74, 69, 63, 58, 54, 50]
neg_trend = [14, 19, 25, 29, 33, 36]
ax9.plot(years, pos_trend, 'o-', color=COLORS['positive'], linewidth=3, label='Positive')
ax9.plot(years, neg_trend, 's-', color=COLORS['negative'], linewidth=3, label='Negative')
ax9.fill_between(years, pos_trend, neg_trend, alpha=0.2, color=COLORS['neutral'])
ax9.set_title('SENTIMENT TREND 2021-2026', fontsize=12, weight='bold')
ax9.legend(); ax9.grid(True, alpha=0.3)

# 10. COMPETITIVE MENTIONS
ax10 = fig.add_subplot(4, 3, 10)
comps = ['Chicken\nRepublic', 'The Place', 'Gourmet\nTwist', 'Kilimanjaro', 'Sweet\nSensation']
cvals = [14, 9, 5, 4, 6]
ccols = ['#FF5722', '#4CAF50', '#2196F3', '#9C27B0', '#FF9800']
ax10.bar(comps, cvals, color=ccols)
ax10.set_title('COMPETITIVE MENTIONS', fontsize=12, weight='bold')

# 11. EXPANSION DEMAND
ax11 = fig.add_subplot(4, 3, 11)
ecities = ['Abuja', 'Port\nHarcourt', 'Ibeju\nLekki', 'Akure', 'Yaba', 'Ibadan']
edemand = [96, 48, 38, 22, 32, 18]
ax11.bar(ecities, edemand, color=COLORS['anticipation'])
ax11.set_title('EXPANSION DEMAND BY CITY', fontsize=12, weight='bold')

# 12. BRAND HEALTH SCORECARD
ax12 = fig.add_subplot(4, 3, 12)
bhcats = ['Food\nQuality', 'Service', 'Value', 'Hygiene', 'Trust', 'Innovation', 'Accessibility']
bhscores = [73, 36, 59, 81, 66, 46, 32]
bhcolors = [COLORS['positive'] if s>=60 else COLORS['negative'] if s<40 else COLORS['warning'] for s in bhscores]
bars = ax12.barh(bhcats, bhscores, color=bhcolors)
ax12.set_xlim(0, 100); ax12.axvline(x=60, color='gray', linestyle='--', alpha=0.5)
ax12.set_title('BRAND HEALTH SCORECARD (Target: 60+)', fontsize=12, weight='bold')
for b, v in zip(bars, bhscores):
    ax12.text(b.get_width()+1, b.get_y()+b.get_height()/2, f'{v}', va='center', weight='bold')

plt.suptitle('MEGA CHICKEN RESTAURANT\nCORPORATE BRAND INTELLIGENCE DASHBOARD', 
             fontsize=20, weight='bold', y=0.995)
plt.tight_layout(rect=[0, 0, 1, 0.99])

# SAVE FILES
plt.savefig('MEGA_CHICKEN_DASHBOARD.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.savefig('MEGA_CHICKEN_DASHBOARD.pdf', dpi=300, bbox_inches='tight', facecolor='white')
print("Step 4/5: Dashboard saved as PNG and PDF!")
plt.show()

print("\n✅ ALL DONE! Check your folder for MEGA_CHICKEN_DASHBOARD.png and .pdf")
```
